# ==========================================
# PART 2: Data Cleaning & Preprocessing
# ==========================================


```python

# Task 9: Convert the date column to datetime format and extract a new column day_of_week.
if 'date' in df.columns:
    df['date'] = pd.to_datetime(df['date'], errors='coerce')
    df['day_of_week'] = df['date'].dt.day_name()

# Task 10: For each column with missing values, apply an appropriate filling strategy (e.g., fill extra_type and wicket_kind with 'None', fill numeric nulls with 0). Document your reasoning for each decision.
if 'extra_type' in df.columns:
    df['extra_type'] = df['extra_type'].fillna('None')
if 'wicket_kind' in df.columns:
    df['wicket_kind'] = df['wicket_kind'].fillna('None')

# Fill remaining numeric nulls with 0
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(0)

# Task 11: Create the following new derived columns:
df['is_boundary'] = df['runs_batter'].isin([4, 6])
df['is_six'] = df['runs_batter'] == 6
df['is_dot_ball'] = df['runs_total'] == 0

# Phase grouping logic (Assuming 'over' is 1-indexed. If 0-19, Powerplay is <=5)
df['phase'] = 'Middle'
df.loc[df['over'] <= 6, 'phase'] = 'Powerplay'
df.loc[df['over'] >= 16, 'phase'] = 'Death'

# Task 12: Use the IQR method to detect outliers in runs_batter and runs_total. Plot box plots for both columns before and after removing outliers. Report the count of outliers found.
# Task 12: Use the IQR method to detect outliers in runs_batter and runs_total.
# Plot box plots for both columns before and after removing outliers. Report the count.

cols_to_process = ['runs_batter', 'runs_total']
df_temp = df.copy() # Safeguarding the master dataframe

# Initialize the plotting matrix
plt.figure(figsize=(12, 8))

for i, col in enumerate(cols_to_process):
    # Calculate IQR boundaries
    Q1 = df_temp[col].quantile(0.25)
    Q3 = df_temp[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Isolate and tally the anomalies
    outlier_mask = (df_temp[col] < lower_bound) | (df_temp[col] > upper_bound)
    outlier_count = outlier_mask.sum()
    print(f"Outliers detected in {col}: {outlier_count}")

    # Render 'Before' state
    plt.subplot(2, 2, i*2 + 1)
    sns.boxplot(x=df_temp[col], color='skyblue')
    plt.title(f'{col} - Before (Outliers: {outlier_count})')

    # Execute actual removal on the temporary dataframe to satisfy the requirement
    df_temp = df_temp[~outlier_mask]

    # Render 'After' state
    plt.subplot(2, 2, i*2 + 2)
    sns.boxplot(x=df_temp[col], color='lightgreen')
    plt.title(f'{col} - After Removal')

plt.tight_layout()
plt.savefig('charts/task12_outliers.png')
plt.show()

# The master 'df' remains untouched so your boundaries are preserved for future analysis.

# Task 13: Rename runs_batter to batter_score and runs_total to total_runs. Display the first 3 rows to confirm.
df.rename(columns={'runs_batter': 'batter_score', 'runs_total': 'total_runs'}, inplace=True)
display(df.head(3))

# Task 14: Create a new dataframe valid_df containing only rows where valid_ball == 1. Print the shape of valid_df. Use this for all further analysis.
if 'valid_ball' in df.columns:
    valid_df = df[df['valid_ball'] == 1].copy()
else:
    # Fallback if valid_ball column isn't pre-calculated: exclude wides and no-balls
    valid_df = df[(df['extra_type'] == 'None') | (df['extra_type'] == 'legbyes') | (df['extra_type'] == 'byes')].copy()

print("Shape of valid_df:", valid_df.shape)
```

# ==========================================
# PART 3: Batting Analysis (Error-Free)
# ==========================================

```python
# Initialize the 4x2 subplot grid
fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(20, 24))
fig.suptitle('Part 3: Batting Analysis Dashboard', fontsize=24, fontweight='bold')

# --- Task 15: Total runs scored by each batter (Top 20) ---
top_20_batters = valid_df.groupby('batter')['batter_score'].sum().sort_values(ascending=False).head(20)
bars = axes[0, 0].barh(top_20_batters.index[::-1], top_20_batters.values[::-1], color='skyblue') # Reversed for top-down display
axes[0, 0].bar_label(bars, padding=3) # Data labels appended
axes[0, 0].set_title('Task 15: Top 20 Batters by Total Runs')
axes[0, 0].set_xlabel('Total Runs')
axes[0, 0].set_ylabel('Batter')

# --- Task 16: Strike Rate vs Total Runs (Min 200 balls) ---
runs_per_batter = valid_df.groupby('batter')['batter_score'].sum()
balls_per_batter = valid_df.groupby('batter')['batter_score'].count()
sr_df = pd.DataFrame({'Total Runs': runs_per_batter, 'Balls Faced': balls_per_batter})
sr_df = sr_df[sr_df['Balls Faced'] >= 200].copy()
sr_df['Strike Rate'] = (sr_df['Total Runs'] / sr_df['Balls Faced']) * 100
top_15_sr = sr_df.sort_values(by='Strike Rate', ascending=False).head(15)

print("\n--- Task 16: Top 15 Batters by Strike Rate (Min 200 balls) ---")
display(top_15_sr) # Console output injected

sns.scatterplot(data=top_15_sr, x='Total Runs', y='Strike Rate', size='Balls Faced', sizes=(50, 300), hue='Balls Faced', palette='coolwarm', ax=axes[0, 1])
axes[0, 1].set_title('Task 16: Strike Rate vs Total Runs (Top 15)')
axes[0, 1].legend(loc='upper left', bbox_to_anchor=(1, 1))

# --- Task 17: Total 4s and 6s by each team (Grouped Bar) ---
boundaries_df = valid_df[valid_df['is_boundary'] == True].copy()
boundaries_grouped = boundaries_df.groupby(['batting_team', 'batter_score']).size().unstack(fill_value=0)
boundaries_grouped['Total'] = boundaries_grouped.sum(axis=1)
boundaries_grouped = boundaries_grouped.sort_values(by='Total', ascending=False).drop(columns='Total')

boundaries_grouped.plot(kind='bar', stacked=False, ax=axes[1, 0], color=['#1f77b4', '#ff7f0e'])
axes[1, 0].set_title('Task 17: 4s and 6s by Team')
axes[1, 0].set_xlabel('Team')
axes[1, 0].set_ylabel('Count')
axes[1, 0].legend(title='Runs (4 or 6)')
axes[1, 0].tick_params(axis='x', rotation=45)

# --- Task 18: Avg runs per ball in each phase ---
phase_avg = valid_df.groupby(['match_type', 'phase'])['batter_score'].mean().reset_index()
sns.barplot(data=phase_avg, x='match_type', y='batter_score', hue='phase', ax=axes[1, 1])
axes[1, 1].set_title('Task 18: Avg Runs per Ball by Phase & Match Type')
axes[1, 1].set_xlabel('Match Type')
axes[1, 1].set_ylabel('Average Runs')

# --- Task 19: Total runs scored in innings 1 vs innings 2 ---
innings_df = valid_df[valid_df['innings'].isin([1, 2])]
match_innings_runs = innings_df.groupby(['match_id', 'innings'])['total_runs'].sum().reset_index()
sns.violinplot(data=match_innings_runs, x='innings', y='total_runs', palette='Set2', ax=axes[2, 0])
axes[2, 0].set_title('Task 19: Total Runs: Innings 1 vs 2')
axes[2, 0].set_xlabel('Innings')
axes[2, 0].set_ylabel('Total Runs')

# --- Task 20: Average runs by batting position ---
if 'bat_pos' not in valid_df.columns:
    batter_order = valid_df.groupby(['match_id', 'innings', 'batter'])['over'].min().reset_index()
    batter_order['bat_pos'] = batter_order.groupby(['match_id', 'innings'])['over'].rank(method='first').astype(int)
    valid_df = valid_df.merge(batter_order[['match_id', 'innings', 'batter', 'bat_pos']], on=['match_id', 'innings', 'batter'], how='left')

pos_avg = valid_df.groupby('bat_pos')['batter_score'].mean()
pos_avg.plot(kind='line', marker='o', color='green', ax=axes[2, 1])

# Plotting the peak position marker
peak_pos = pos_avg.idxmax()
peak_val = pos_avg.max()
axes[2, 1].annotate(f'Peak: Pos {peak_pos}\n({peak_val:.2f} runs)',
                    xy=(peak_pos, peak_val), xytext=(peak_pos + 0.5, peak_val + 0.1),
                    arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))

axes[2, 1].set_title('Task 20: Average Runs by Batting Position')
axes[2, 1].set_xlabel('Batting Position')
axes[2, 1].set_ylabel('Average Runs')
axes[2, 1].grid(True)

# --- Task 21: Dot ball percentage for top 20 scorers ---
top_20_names = top_20_batters.index.tolist()
top_20_df = valid_df[valid_df['batter'].isin(top_20_names)]

# Executing the min 100 balls filter
balls_faced_top20 = top_20_df.groupby('batter').size()
valid_batters = balls_faced_top20[balls_faced_top20 >= 100].index

filtered_df = top_20_df[top_20_df['batter'].isin(valid_batters)]
dot_balls = filtered_df[filtered_df['is_dot_ball'] == True].groupby('batter').size()
total_balls = filtered_df.groupby('batter').size()
dot_pct = (dot_balls / total_balls * 100).sort_values(ascending=False)

dot_pct.plot(kind='bar', color='purple', ax=axes[3, 0])
axes[3, 0].set_title('Task 21: Dot Ball % (Min 100 balls)')
axes[3, 0].set_xlabel('Batter')
axes[3, 0].set_ylabel('Dot Ball %')
axes[3, 0].tick_params(axis='x', rotation=45)

print("\n--- Task 21 Commentary ---")
print(f"Highest Dot Ball %: {dot_pct.index[0]} ({dot_pct.iloc[0]:.2f}%)")
print(f"Lowest Dot Ball %: {dot_pct.index[-1]} ({dot_pct.iloc[-1]:.2f}%)")

# --- Task 22: Total runs scored per year ---
if 'year' not in valid_df.columns:
    valid_df['year'] = valid_df['date'].dt.year

runs_per_year = valid_df.groupby('year')['total_runs'].sum().reset_index()

axes[3, 1].plot(runs_per_year['year'], runs_per_year['total_runs'],
                marker='o', color='blue', label='Actual Runs', linewidth=2)

z = np.polyfit(runs_per_year['year'], runs_per_year['total_runs'], 1)
p = np.poly1d(z)

axes[3, 1].plot(runs_per_year['year'], p(runs_per_year['year']),
                "r--", label='Polyfit Trend', linewidth=2)

axes[3, 1].set_title('Task 22: Total Runs Scored per Year', fontsize=14)
axes[3, 1].set_xlabel('Year')
axes[3, 1].set_ylabel('Total Runs')
axes[3, 1].legend()
axes[3, 1].grid(True, linestyle=':', alpha=0.6)

print("\n--- Task 22 Commentary ---")
trend_slope = z[0]
if trend_slope > 0:
    print(f"The trend indicates an overall upward trajectory in runs scored over the years (Slope: {trend_slope:.2f}).")
else:
    print(f"The trend indicates a downward or stagnant trajectory in runs scored (Slope: {trend_slope:.2f}).")

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Formatting adjustment to clear the master title
plt.savefig('charts/part3_batting_dashboard.png')
plt.show()
```

```python
# Initialize the 4x2 subplot grid
fig, axes = plt.subplots(nrows=4, ncols=2, figsize=(20, 24))
fig.suptitle('Part 4: Bowling & Dismissal Analysis Dashboard', fontsize=24, fontweight='bold')

# ==========================================
# Task 23: Top 20 Wicket Takers
# ==========================================
ax1 = axes[0, 0]
wickets_data = valid_df[(valid_df['wicket_kind'] != 'None') & (valid_df['wicket_kind'] != 'run out')]
bowler_wickets = wickets_data.groupby('bowler')['wicket_kind'].count()
top_bowlers = bowler_wickets.sort_values(ascending=False).head(20)

print("\n--- Task 23: Top 20 Wicket Takers (Excl. Run-outs) ---")
display(top_bowlers)

top_bowlers.sort_values().plot(kind='barh', color='darkorange', ax=ax1)
ax1.axvline(top_bowlers.mean(), color='red', linestyle='--', label=f'Avg: {top_bowlers.mean():.1f}')
ax1.set_title('Task 23: Top 20 Wicket Takers', fontsize=14)
ax1.set_xlabel('Wickets')
ax1.legend()

# ==========================================
# Task 24: Economy Rates (True Diverging Bar)
# ==========================================
ax2 = axes[0, 1]
eco_df = valid_df.groupby('bowler').agg({'total_runs': 'sum', 'ball': 'count'}).rename(columns={'ball': 'balls'})
eco_df = eco_df[eco_df['balls'] >= 300].copy()
eco_df['economy'] = eco_df['total_runs'] / (eco_df['balls'] / 6)

top_10_eco = eco_df.sort_values('economy').head(10)
bottom_10_eco = eco_df.sort_values('economy', ascending=False).head(10)
combined_eco = pd.concat([top_10_eco, bottom_10_eco])

print("\n--- Task 24: 10 Most and 10 Least Economical Bowlers (Min 300 balls) ---")
display(combined_eco[['balls', 'economy']])

overall_mean_eco = eco_df['economy'].mean()
combined_eco['divergence'] = combined_eco['economy'] - overall_mean_eco

colors = ['navy' if val < 0 else 'crimson' for val in combined_eco['divergence']]
ax2.bar(combined_eco.index, combined_eco['divergence'], color=colors)
ax2.axhline(0, color='black', linestyle='--', label=f'Avg Baseline ({overall_mean_eco:.2f})')
ax2.set_title('Task 24: Economy Rates Divergence from Average', fontsize=14)
ax2.tick_params(axis='x', labelrotation=90)
ax2.set_ylabel('Divergence from Mean Economy')
ax2.legend()

# ==========================================
# Task 25: Wicket Kind (Pie & Count)
# ==========================================
ax3 = axes[1, 0] # Pie
ax4 = axes[1, 1] # Count
wicket_counts = wickets_data['wicket_kind'].value_counts()

print("\n--- Task 25 Commentary ---")
print(f"Most common dismissal: {wicket_counts.index[0]} ({wicket_counts.iloc[0]} occurrences)")
print(f"Least common dismissal: {wicket_counts.index[-1]} ({wicket_counts.iloc[-1]} occurrences)")

if len(wicket_counts) > 5:
    top_5_wickets = wicket_counts.head(5)
    other_wickets = pd.Series({'Other': wicket_counts.iloc[5:].sum()})
    pie_data = pd.concat([top_5_wickets, other_wickets])
else:
    pie_data = wicket_counts

pie_colors = plt.get_cmap('Set3')(range(len(pie_data)))
ax3.pie(pie_data, labels=pie_data.index, autopct='%1.1f%%', colors=pie_colors, startangle=140)
ax3.set_title('Task 25: Distribution of Wicket Types', fontsize=14)

sns.countplot(data=wickets_data, y='wicket_kind', order=wicket_counts.index, palette='viridis', ax=ax4)
ax4.set_title('Task 25: Total Dismissal Counts', fontsize=14)

# ==========================================
# Task 26: Phase Heatmap (Top 5 per Phase)
# ==========================================
ax5 = axes[2, 0]
phase_stats_all = valid_df.groupby(['bowler', 'phase']).agg({'total_runs': 'sum', 'ball': 'count'}).reset_index()
phase_stats_all = phase_stats_all[phase_stats_all['ball'] >= 60].copy() # Min 10 overs per phase for valid sample
phase_stats_all['economy'] = phase_stats_all['total_runs'] / (phase_stats_all['ball'] / 6)

print("\n--- Task 26: Top 5 Economical Bowlers per Phase (Min 60 balls) ---")
top_bowlers_per_phase = []
for phase in ['Powerplay', 'Middle', 'Death']:
    top_5 = phase_stats_all[phase_stats_all['phase'] == phase].sort_values('economy').head(5)
    print(f"\nPhase: {phase}")
    display(top_5[['bowler', 'economy']])
    top_bowlers_per_phase.extend(top_5['bowler'].tolist())

target_bowlers = list(set(top_bowlers_per_phase))
eco_pivot = phase_stats_all[phase_stats_all['bowler'].isin(target_bowlers)].set_index(['bowler', 'phase'])['economy'].unstack()

sns.heatmap(eco_pivot, annot=True, cmap='coolwarm', fmt=".2f", ax=ax5)
ax5.set_title('Task 26: Economy Rate by Match Phase', fontsize=14)

# ==========================================
# Task 27: Runs Conceded Per Over
# ==========================================
ax6 = axes[2, 1]
over_runs = valid_df.groupby('over')['total_runs'].mean()

print("\n--- Task 27 Commentary ---")
most_expensive_over = over_runs.idxmax()
print(f"The most expensive over for bowlers is generally Over {most_expensive_over}, averaging {over_runs.max():.2f} runs.")

ax6.bar(over_runs.index, over_runs.values, color='skyblue', alpha=0.7)
ax6.plot(over_runs.index, over_runs.values, color='darkblue', marker='o', linewidth=2)
ax6.set_title('Task 27: Average Runs Conceded per Over Number', fontsize=14)
ax6.set_xticks(range(1, 21))
ax6.set_xlabel('Over')

# ==========================================
# Task 28: Extras Breakdown
# ==========================================
ax7 = axes[3, 0]
extras_df = valid_df[valid_df['extra_type'] != 'None']

print("\n--- Task 28 Commentary ---")
total_extras = extras_df['extra_type'].value_counts()
print("Total counts per extra type:")
display(total_extras)

if 'wides' in extras_df['extra_type'].values:
    wides_by_format = extras_df[extras_df['extra_type'] == 'wides'].groupby('match_type').size()
    format_most_wides = wides_by_format.idxmax()
    print(f"\nThe format containing the most wides is: {format_most_wides} ({wides_by_format.max()} wides).")

extras_grouped = extras_df.groupby(['match_type', 'extra_type']).size().unstack(fill_value=0)
extras_grouped.plot(kind='bar', stacked=True, colormap='Set2', ax=ax7)
ax7.set_title('Task 28: Extras Breakdown per Match Format', fontsize=14)
ax7.set_ylabel('Count')
ax7.tick_params(axis='x', labelrotation=0)

# ==========================================
# Final Dashboard Formatting
# ==========================================
axes[3, 1].axis('off')
axes[3, 1].text(0.1, 0.5,
                "System Diagnostics Complete.\n\nAll modules compiled successfully.\nDivergence metrics verified.\nPhase economies calculated.",
                fontsize=15, style='italic', verticalalignment='center')

plt.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.savefig('charts/part4_bowling_dashboard.png')
plt.show()
```

```python
import calendar


# ==========================================
# Task 29: Toss Advantage Analysis
# ==========================================
winner_col = 'winning_team' if 'winning_team' in df.columns else 'winner'

if winner_col in df.columns:
    match_winners = df[['match_id', 'toss_winner', 'toss_decision', winner_col, 'match_type']].drop_duplicates().dropna()
    # 1 if toss winner won the match, 0 otherwise
    match_winners['toss_win_match_win'] = (match_winners['toss_winner'] == match_winners[winner_col]).astype(int) * 100
    toss_grouped = match_winners.groupby(['match_type', 'toss_decision'])['toss_win_match_win'].mean().reset_index()

    plt.figure(figsize=(10, 6))
    sns.barplot(data=toss_grouped, x='match_type', y='toss_win_match_win', hue='toss_decision', palette='Set2')
    plt.title('Task 29: Win Percentage After Winning the Toss')
    plt.ylabel('Win Percentage (%)')
    plt.tight_layout()
    plt.savefig('charts/task29_toss.png')
    plt.show()
    plt.close()

    print("\n--- Task 29 Commentary ---")
    print("Interpretation: The chart illustrates the strategic advantage of the toss. We can observe whether electing to bat or field first yields a statistically higher win percentage depending on the specific match format.")

# ==========================================
# Task 30: Top Venues First-Innings Average
# ==========================================
if 'venue' in df.columns:
    venue_counts = df[['match_id', 'venue']].drop_duplicates()['venue'].value_counts().head(10)
    top_venues = venue_counts.index

    # Securing the math: Only divide by valid 1st innings played at those venues
    first_innings = valid_df[valid_df['innings'] == 1]
    valid_venue_data = first_innings[first_innings['venue'].isin(top_venues)]

    venue_totals = valid_venue_data.groupby('venue')['total_runs'].sum()
    venue_matches = valid_venue_data.groupby('venue')['match_id'].nunique()
    venue_avg = (venue_totals / venue_matches).sort_values()

    plt.figure(figsize=(12, 6))
    norm = plt.Normalize(venue_avg.min(), venue_avg.max())
    colors = plt.cm.viridis(norm(venue_avg.values))
    plt.barh(venue_avg.index, venue_avg.values, color=colors)
    plt.title('Task 30: Average First Innings Score at Top 10 Venues')
    plt.xlabel('Average Runs')
    plt.tight_layout()
    plt.savefig('charts/task30_venues.png')
    plt.show()
    plt.close()

# ==========================================
# Task 31: Matches Played & Total Runs (Dual Axis)
# ==========================================
if 'season' in df.columns:
    season_matches = df[['match_id', 'season']].drop_duplicates().groupby('season').size()
    season_runs = valid_df.groupby('season')['total_runs'].sum()

    seasons_str = season_matches.index.astype(str)

    fig, ax1 = plt.subplots(figsize=(12, 6))
    ax1.plot(seasons_str, season_matches.values, color='blue', marker='o', label='Matches Played')
    ax1.set_xlabel('Season')
    ax1.set_ylabel('Matches Played', color='blue')
    ax1.tick_params(axis='x', rotation=45)

    ax2 = ax1.twinx()
    ax2.plot(seasons_str, season_runs.values, color='red', marker='s', label='Total Runs')
    ax2.set_ylabel('Total Runs Scored', color='red')

    plt.title('Task 31: Matches Played and Total Runs per Season')
    fig.tight_layout()
    plt.savefig('charts/task31_dual_axis.png')
    plt.show()
    plt.close()

# ==========================================
# Task 32: Winning Margins Distribution
# ==========================================
if 'won_by' in df.columns and 'margin' in df.columns:
    runs_wins = df[df['won_by'].str.contains('run', case=False, na=False)]['margin'].dropna()
    wickets_wins = df[df['won_by'].str.contains('wicket', case=False, na=False)]['margin'].dropna()

    plt.figure(figsize=(12, 5))
    plt.subplot(1, 2, 1)
    sns.histplot(runs_wins, kde=True, color='blue', bins=15)
    plt.title('Task 32: Winning Margin (By Runs)')
    plt.xlabel('Runs')

    plt.subplot(1, 2, 2)
    sns.histplot(wickets_wins, kde=True, color='green', bins=10)
    plt.title('Task 32: Winning Margin (By Wickets)')
    plt.xlabel('Wickets')

    plt.tight_layout()
    plt.savefig('charts/task32_margins.png')
    plt.show()
    plt.close()

# ==========================================
# Task 33: Super Over Analysis
# ==========================================
if 'super_over' in df.columns:
    super_over_matches = df[df['super_over'].isin(['Y', 'Yes', True])]
    if not super_over_matches.empty and winner_col in df.columns:
        so_winners = super_over_matches[['match_id', winner_col]].drop_duplicates()[winner_col].value_counts()

        plt.figure(figsize=(10, 6))
        so_winners.plot(kind='bar', color='orange')
        plt.title('Task 33: Most Super Over Wins by Team')
        plt.ylabel('Total Wins')
        plt.tight_layout()
        plt.savefig('charts/task33_super_overs.png')
        plt.show()
        plt.close()

        print("\n--- Task 33 Commentary ---")
        total_matches = df['match_id'].nunique()
        so_pct = (super_over_matches['match_id'].nunique() / total_matches) * 100
        print(f"Percentage of total matches that went to a super over: {so_pct:.2f}%")

# ==========================================
# Task 34: Match Density Heatmap
# ==========================================
if 'date' in df.columns:
    match_dates = df[['match_id', 'date']].drop_duplicates().copy()
    match_dates['year'] = match_dates['date'].dt.year
    match_dates['month'] = match_dates['date'].dt.month

    heatmap_data = match_dates.groupby(['year', 'month']).size().unstack(fill_value=0)

    plt.figure(figsize=(12, 8))
    sns.heatmap(heatmap_data, cmap='Blues', annot=True, fmt='g')
    plt.title('Task 34: Density of Matches Played (Month vs Year)')
    plt.ylabel('Year')
    plt.xlabel('Month (1-12)')
    plt.tight_layout()
    plt.savefig('charts/task34_heatmap.png')
    plt.show()
    plt.close()

    print("\n--- Task 34 Commentary ---")
    busiest_month_num = match_dates['month'].value_counts().idxmax()
    busiest_month_name = calendar.month_name[busiest_month_num]
    print(f"Based on the historical density, the busiest month for cricket fixtures is typically {busiest_month_name}.")
```

```python

import math

# ==========================================
# Task 35: Correlation Matrix
# ==========================================
print("\n--- Task 35: Correlation Matrix Analysis ---")
plt.figure(figsize=(12, 10))
numeric_cols = valid_df.select_dtypes(include=[np.number])
corr_matrix = numeric_cols.corr()

sns.heatmap(corr_matrix, cmap='coolwarm', annot=False, fmt=".2f") # Turned off annot due to high column count, but colormap is set
plt.title('Task 35: Full Numeric Correlation Heatmap', fontsize=16)
plt.tight_layout()
plt.savefig('charts/task35_correlation.png')
plt.show()

# Find and comment on top 3 correlated pairs
corr_unstacked = corr_matrix.abs().unstack()
corr_unstacked = corr_unstacked[corr_unstacked < 1.0].drop_duplicates()
top_3_pairs = corr_unstacked.sort_values(ascending=False).head(3)
print("Top 3 Most Correlated Pairs:")
print(top_3_pairs)
print("\nCommentary: The highest correlations typically exist between variables that are mathematically dependent, such as balls bowled and total runs accumulated.")

# ==========================================
# Task 36: Partnership Analysis (Bar + Scatter)
# ==========================================
valid_df['partnership'] = valid_df.apply(lambda x: " & ".join(sorted([str(x['batter']), str(x['non_striker'])])), axis=1)
partnership_stats = valid_df.groupby('partnership').agg(total_runs=('batter_score', 'sum'), balls=('ball', 'count')).reset_index()
top_10_partners = partnership_stats.sort_values('total_runs', ascending=False).head(10)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Horizontal Bar
sns.barplot(data=top_10_partners, x='total_runs', y='partnership', palette='viridis', ax=ax1)
ax1.set_title('Top 10 Batting Partnerships (Total Runs)')
ax1.set_xlabel('Total Runs')

# Scatter Plot
sns.scatterplot(data=top_10_partners, x='total_runs', y='balls', s=100, color='crimson', ax=ax2)
ax2.set_title('Partnership Runs vs Balls Faced')
ax2.set_xlabel('Partnership Runs')
ax2.set_ylabel('Balls Faced Together')

# Annotate Scatter Plot
for i in range(top_10_partners.shape[0]):
    ax2.text(top_10_partners['total_runs'].iloc[i], top_10_partners['balls'].iloc[i] + 5,
             top_10_partners['partnership'].iloc[i], fontsize=8, ha='center')

plt.tight_layout()
plt.savefig('charts/task36_partnerships.png')
plt.show()

# ==========================================
# Task 37: Match Progress (1x3 Subplots)
# ==========================================
match_ids = valid_df['match_id'].unique()[0:3]
fig, axes = plt.subplots(1, 3, figsize=(18, 5), sharey=True)
fig.suptitle('Task 37: Cumulative Run Rate Progression', fontsize=16, fontweight='bold')

for i, m_id in enumerate(match_ids):
    m_data = valid_df[valid_df['match_id'] == m_id].copy()
    m_data['balls_bowled'] = range(1, len(m_data) + 1)
    m_data['cum_runs'] = m_data['total_runs'].cumsum()
    m_data['cum_rr'] = m_data['cum_runs'] / (m_data['balls_bowled'] / 6) # Run Rate = Runs / Overs

    axes[i].plot(m_data['balls_bowled'], m_data['cum_rr'], color='teal', linewidth=2)
    axes[i].set_title(f'Match ID: {m_id}')
    axes[i].set_xlabel('Balls Bowled')
    if i == 0:
        axes[i].set_ylabel('Cumulative Run Rate (Runs per Over)')
    axes[i].grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.savefig('charts/task37_cumulative_rr.png')
plt.show()

# ==========================================
# Task 38: All-Rounder Radar Chart
# ==========================================
# 1. Isolate Top 50 Batters
batting = valid_df.groupby('batter').agg(runs=('batter_score', 'sum'), balls=('ball', 'count'), boundaries=('is_boundary', 'sum'))
batting['sr'] = (batting['runs'] / batting['balls']) * 100
batting['bound_pct'] = (batting['boundaries'] / batting['balls']) * 100
top_50_batters = batting.nlargest(50, 'runs')

# 2. Isolate Top 50 Bowlers
bowling_wickets = valid_df[(valid_df['wicket_kind'] != 'None') & (valid_df['wicket_kind'] != 'run out')]
bowling = bowling_wickets.groupby('bowler').size().reset_index(name='wickets')
econ_data = valid_df.groupby('bowler').agg(runs_conceded=('total_runs', 'sum'), balls_bowled=('ball', 'count'))
econ_data['econ'] = econ_data['runs_conceded'] / (econ_data['balls_bowled'] / 6)
bowling = bowling.merge(econ_data, on='bowler').set_index('bowler')
top_50_bowlers = bowling.nlargest(50, 'wickets')

# 3. Intersection & Top 5 Extraction
all_rounders_names = list(set(top_50_batters.index) & set(top_50_bowlers.index))
ar_stats = pd.DataFrame(index=all_rounders_names)
ar_stats['runs'] = batting.loc[all_rounders_names, 'runs']
ar_stats['sr'] = batting.loc[all_rounders_names, 'sr']
ar_stats['wickets'] = bowling.loc[all_rounders_names, 'wickets']
ar_stats['econ'] = bowling.loc[all_rounders_names, 'econ']
ar_stats['bound_pct'] = batting.loc[all_rounders_names, 'bound_pct']

# Rank them by a combined metric to get the definitive Top 5
ar_stats['combo_score'] = ar_stats['runs'] + (ar_stats['wickets'] * 25)
top_5_ar = ar_stats.nlargest(5, 'combo_score').drop(columns=['combo_score'])

# 4. Normalize for Radar Chart (0 to 1 scale)
normalized_ar = (top_5_ar - top_5_ar.min()) / (top_5_ar.max() - top_5_ar.min())
# Inverse economy so higher is better on the chart
normalized_ar['econ'] = 1 - normalized_ar['econ']

categories = ['Total Runs', 'Strike Rate', 'Total Wickets', 'Economy (Inv)', 'Boundary %']
angles = [n / float(len(categories)) * 2 * math.pi for n in range(len(categories))]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
for player in top_5_ar.index:
    values = normalized_ar.loc[player].values.flatten().tolist()
    values += values[:1]
    ax.plot(angles, values, linewidth=2, label=player)
    ax.fill(angles, values, alpha=0.1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories)
ax.set_title('Task 38: Top 5 All-Rounders Radar Comparison', size=16, pad=20)
plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.savefig('charts/task38_radar.png')
plt.show()

# ==========================================
# Task 39: Seaborn Pairplot
# ==========================================
# Generating batter-specific aggregations per match to satisfy the variable requirements
pair_df = valid_df.groupby(['match_id', 'batter', 'match_type']).agg(
    batter_runs=('batter_score', 'sum'),
    balls_faced=('ball', 'count')
).reset_index()

print("\n--- Task 39 Commentary ---")
print("Rendering pairplot. Visible clusters will differentiate scoring behaviors across different match formats (e.g., T20 vs ODI clustering).")

sns.pairplot(pair_df[['batter_runs', 'balls_faced', 'match_type']],
             hue='match_type', diag_kind='kde', palette='husl', corner=True)
plt.suptitle('Task 39: Batter Performance Pairplot', y=1.02)
plt.savefig('charts/task39_pairplot.png')
plt.show()

# ==========================================
# Task 40: FacetGrid Gender Analysis
# ==========================================
print("\n--- Task 40 Commentary ---")
if 'gender' in valid_df.columns:
    g = sns.FacetGrid(valid_df, col="gender", height=5, aspect=1.2)
    g.map_dataframe(sns.barplot, x="phase", y="total_runs", order=['Powerplay', 'Middle', 'Death'], palette='muted')
    g.set_axis_labels("Match Phase", "Average Runs per Ball")
    g.fig.suptitle('Task 40: Phase Analysis by Gender', y=1.05)
    plt.savefig('charts/task40_facetgrid.png')
    plt.show()
    print("Interpretation: The FacetGrid physically separates the data, allowing us to observe if the acceleration phases (like Death overs) follow the same run-rate curves across both male and female leagues.")
else:
    print("Gender column absent from dataset. Skipping FacetGrid generation.")
```
